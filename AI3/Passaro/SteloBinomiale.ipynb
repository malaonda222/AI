```python
#ESEMPIO BINOMIALE CON PMF = indichiamo tutte le probabilità possibili della variabile discreta

#Memo codice per grafico a stelo Binomiale 
#al variare di p puoi vedere come "cambia" il grafico
#con p=0.5 hai un grafico simmetrico
#con p=0.2 hai un grafico molto asimmetrico 

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom

#Parametri della binomiale 
n = 10
p = 0.2

#Valori possibili della variabile casuale 
x = np.arange(0, n+1)

#Probabilità P(X = k)
pmf = binom.pmf(x, n, p)

#Stem plot
plt.stem(x, pmf, basefmt=" ") #basefmt=" " toglie riga rossa
plt.xlabel('k')
plt.ylabel('P(X = k)')
plt.title('Distribuzione binomiale (n=10, p=0.2)')
plt.grid(True)
plt.show()
```

```python
#ESEMPIO BINOMIALE CON RVS = generiamo molti valori casuali secondo quella distribuzione e osserviamo
#come si distribuiscono nella pratica

#altro esempio 
#mostro istogrammi che fanno vedere che la distribuzione binomiale non è simmetrica
#con p=0.2, 0.3, 0.5

import numpy as np 
import matplotlib.pyplot as plt
from scipy.stats import binom

#Parametri 
n = 10
p = 0.2 
N = 10000 #numero di simulazioni

#Campionamento della binomiale 
samples = binom.rvs(n, p, size=N)

#Istogramma
plt.hist(samples, bins=np.arange(-0.5, n + 1.5, 1), density=True)
plt.xlabel('k')
plt.ylabel('Frequenza relativa')
plt.title('Istogramma della Binomiale (n=10, p=0.2)')
plt.grid(True)

plt.show()
```
