```python
import scipy as sp
import numpy as np
import matplotlib.pyplot as plt

print(sp.stats.norm.cdf(1.96)) #se non gli diamo nulla, lui assume che la media sia 0 e la devstand sia 1

#si prende un numero e lo normalizziamo usando la tabella

print(sp.stats.norm.cdf(-2))
```

```python
print(sp.stats.norm.ppf(0.975))

print() 

alpha = 0.05
print(alpha)
print(alpha/2)
print(1-alpha/2)
print(sp.stats.norm.ppf(1-alpha/2))
```

```python
print(sp.stats.norm.cdf(-2))
print(1-sp.stats.norm.cdf(2))
```

```python
#cosa fa CDF? calcola la probabilità di avere un valore <= Z
#cosa fa la PPF? ti da la probabilità e ti calcola il valore di Z che ti dà la probabilità
```

```python
plt.figure(1)
plt.title("Grafico pdf: probability density function")
plt.plot(np.arange(-3, 3, 0.01), sp.stats.norm.pdf(np.arange(-3, 3, 0.01)))
plt.axvline(1.96, color='r', linestyle='--')

plt.figure(2)
plt.title("Grafico cdf: probability density function")
plt.plot(np.arange(-3, 3, 0.01), sp.stats.norm.pdf(np.arange(-3, 3, 0.01)))
plt.axhline(0.975, color='r', linestyle='--')
plt.axvline(1.96, color='r', linestyle='--')

plt.figure(3)
plt.title("Grafico ppf: percent point function")
plt.plot(np.arange(0, 1, 0.01), sp.stats.norm.ppf(np.arange(0, 1, 0.01)))
plt.axvline(0.975, color='r', linestyle='--')
plt.axhline(1.96, color='r', linestyle='--')
```
