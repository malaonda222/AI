```python
# Esercizio 1
# Scrivi il codice di Aggiornamento Bayesiano sequenziale
# Stima della probabilità di una email di Spam e se supera la soglia predefenita (ad es 0.7)
# scrivi se è o non è una mail di Spam
# Prior: P(spam)= 0.4
# Osserviamo una delle parole in una email ed aggiorniamo la probabilità

# Likelihood P(parola | Spam) e P(parola | Non-Spam)
lik_spam = {
    "win":   0.7,
    "meeting":0.05,
    "offer": 0.6,
    "hello": 0.02,
    "prize": 0.8,
    "project":0.04,
    "update": 0.10,
    "invoice": 0.01
}

lik_nonspam = {
    "win":   0.1,
    "meeting":0.2,
    "offer": 0.05,
    "hello": 0.3,
    "prize": 0.08,
    "project":0.25,
    "update": 0.15,
    "invoice": 0.4
}

# sequenza parole osservate
parole = ["win", "meeting", "project", "offer", "prize", "win"]

# Prior iniziale
p_spam = 0.4         #P(B) 
p_non_spam = 0.6     #1 - P(B)
p_spam_dato_parola = 0.7      #P(A|B)
p_non_spam_dato_parola = 0.3  #1 - P(A|B)


# Esercizio 2
# Applicando il codice dell'esercizio precedente, modifica la sequenza delle parole
# e verifica come varia la probabilità di spam

def teorema_bayes(p_a_dato_b, p_b, p_a):
    return(p_a_dato_b*p_b) / p_a

def teorema_bayes_completo(p_a_dato_b, p_b, p_a_dato_non_b):
    p_non_b = 1-p_b
    p_a = p_a_dato_b*p_b + p_a_dato_non_b*p_non_b
    return teorema_bayes(p_a_dato_b, p_b, p_a)

#Opzione 1
storia_probabilita = [p_spam]

for i, parola in enumerate(parole, 1):
    if parola in lik_spam:
        p_evidenza_spam = p_spam_dato_parola
        p_evidenza_non_spam = p_non_spam_dato_parola
    else:
        p_evidenza_spam = 1 - p_spam_dato_parola
        p_evidenza_non_spam = 1 - p_non_spam_dato_parola

    p_spam = teorema_bayes_completo(p_evidenza_spam, p_spam, p_evidenza_non_spam)
    storia_probabilita.append(p_spam)

    print(f"parole[i]: {parola} -> P(spam) = {p_spam:.2f}")

print(f"\nProbabilità finale: {p_spam} = {p_spam:.2%}")

#Opazione 2
p_spam = 0.4 
storia_probabilita = [p_spam]

for parola in parole:
    p_spam = teorema_bayes_completo(lik_spam[parola], p_spam, lik_nonspam[parola])
    storia_probabilita.append(p_spam)

    print(f"parole[i]: {parola} -> P(spam) = {p_spam:.2f}")

print(f"\nProbabilità finale: {p_spam} = {p_spam:.2%}")

# Esercizio 3
# Scrivi il codice di Aggiornamento Bayesiano sequenziale
# che stima della probabilità che un paziente sia malato o sano in funzione
# del risultato di una serie di esami che danno valore positivo o negativo
# ---------------------------------------------------------
# Scenario clinico: un paziente esegue una serie di test
# ---------------------------------------------------------

print("===================================================")
print()

# Prevalenza (prior): P(Malattia)
p_malattia = 0.02   # 2%  #P(B)
p_sano = 0.98 # 1 - P(B)

# Test diagnostico (uguale per tutti i test)
sensibilita = 0.92               # P(Test+ | Malato) #P(A|B)
specificita = 0.95               # P(Test- | Sano) #P(A|B)
falso_positivo = 1 - specificita # P(Test+ | Sano) #1 - P(A|B)
falso_negativo = 1 - sensibilita # P(Test- | Malato) #1 - P(A|B)

# Sequenza reale dei test effettuati
test = ['+', '-', '+', '+', '-']

# Storia della probabilità da plottare
storia_probabilita = [p_malattia]

for i, valore in enumerate(test, 1):
    if valore == "+":
        p_evidenza_malato = sensibilita
        p_evidenza_sano = falso_positivo 
    else:
        p_evidenza_malato = falso_negativo
        p_evidenza_sano = specificita 

    p_malattia = teorema_bayes_completo(p_evidenza_malato, p_malattia, p_evidenza_sano)
    storia_probabilita.append(p_malattia)

    
    print(f"test[i]: {valore} -> P(malattia) = {p_malattia:.2f}")

print(f"\nProbabilità finale: {p_malattia} = {p_malattia:.2%}")
```
