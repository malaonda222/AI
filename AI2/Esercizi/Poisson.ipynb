```python
from scipy.stats import poisson 
#Esempio 1
#Parametro lambda 
lambda_call = 5 

#1. Probabilità di esattamente 3 chiamate 
prob_3 = poisson.pmf(2, lambda_call)
print('Probabilità di ricevere esattamente 3 chiamate:', prob_3)

#2. Probabilità di almeno 4 chiamate
prob_4 = 1 - poisson.pmf(3, lambda_call)
print('Probabilità di ricevere almeno 4 chiamate:', prob_4)

#3. Media e Varianza 
media = lambda_call 
var = lambda_call 
print('Media', media)
print('Varianza', var)
```

```python
#oppure 
#Esempio 2 
import math 

#Parametri
lambda_call = 5 
k = 3 

#Calcolo della probabilità usando la formula di Poisson 

#1. Probabilità di esattamente 3 chiamate
prob = (lambda_call **k * math.exp (-lambda_call)) / math.factorial(k)
print(f'Probabilità di ricevere esattamente 3 chiamate: {prob:.9f}')

#2. Probabilità di almeno 4 chiamate
prob_2 = 1 - ((lambda_call)**3 * math.exp(-lambda_call)) / math.factorial(3)+\
    (lambda_call ** 2 * math.exp(-lambda_call)) / math.factorial(2)+\
        (lambda_call ** 1 * math.exp(-lambda_call) / math.factorial(1)+\
             (lambda_call ** 0 * math.exp(-lambda_call)) / math.factorial(0))
print(f'Probabilità di ricevere almeno 4 chiamate: {prob_2:.9f}')

#3. Probabilità di esattamente 2 chiamate
prob_3 = (lambda_call ** 2 * math.exp(-lambda_call)) / math.factorial(2)
print(f'Probabilità di ricevere esattamente 2 chiamate: {prob_3:.9f}')
```
