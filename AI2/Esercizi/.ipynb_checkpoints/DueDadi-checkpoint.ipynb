```python
import random 

#Possibili risultati
moneta = ['Testa', 'Croce']

#Simula un lancio 
lancio = random.choice(moneta)
print(lancio)
```

```python
lancio = random.randint(0, 1)
if lancio == 1:
    print('Testa')
else:
    print('Croce')
```

```python
moneta = ['Testa', 'Croce']
lanci = [random.choice(moneta) for _ in range(5)]
print(lanci)
print(f"Numero testa: {lanci.count('Testa')}")
print(f"Numero croce: {lanci.count('Croce')}")
```
