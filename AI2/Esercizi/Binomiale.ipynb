```python
n = 10 #numero di prove 
p = 0.3 #probabilità di successo 
mu = n * p 
var = n * p * (1 - p)
print("Media", mu)
print("Varianza", var)
```

```python
import math

def binomiale(n: int, p: float, k: int):
    n_su_k = math.factorial(n) / (math.factorial(k) * math.factorial(n - k))
    p_alla_k = p ** k 
    insuccesso = (1 - p) ** (n - k)
    return n_su_k * p_alla_k * (insuccesso)

print(binomiale(10, 0.5, 3))
```

```python
import scipy

def binomiale1(n: int, p: float, k: int):
    return scipy.stats.binom.pmf(k, n, p)
    
print(binomiale1(10, 0.5, 3))
```

```python
#oppure 
import scipy.stats as binomial
def binomiale2(n:int, p:float, k:int):
    return binomial.binom.pmf(k, n, p)
    
print(binomiale2(10, 0.5, 3))
```
